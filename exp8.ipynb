```python
import random
# Genetic Algorithm: Applying GA Operators
POPULATION_SIZE = 6
CHROMOSOME_LENGTH = 5
GENERATIONS = 10
CROSSOVER_RATE = 0.8
MUTATION_RATE = 0.1

# Step 1: Initialize population
def create_population():
    return [
        [random.randint(0, 1) for _ in
range(CHROMOSOME_LENGTH)]
        for _ in range(POPULATION_SIZE)
    ]

# Convert binary chromosome to decimal
def decode(chromosome):
    return int("".join(map(str, chromosome)), 2)

# Step 2: Calculate fitness
def fitness(chromosome):
    x = decode(chromosome)
    return x ** 2

# Step 3: Tournament selection
def selection(population):
    candidates = random.sample(population, 3)
    return max(candidates, key=fitness)[:]

# Step 4: Single-point crossover
def crossover(parent1, parent2):
    if random.random() < CROSSOVER_RATE:
        point = random.randint(1,
CHROMOSOME_LENGTH - 1)
        child1 = parent1[:point] + parent2[point:]
        child2 = parent2[:point] + parent1[point:]
        return child1, child2
    return parent1[:], parent2[:]

# Step 5: Bit-flip mutation
def mutation(chromosome):
    for i in range(CHROMOSOME_LENGTH):
        if random.random() < MUTATION_RATE:
            chromosome[i] = 1 - chromosome[i]
    return chromosome

# Step 6: Run Genetic Algorithm
population = create_population()

print("Initial Population:")
for individual in population:
    print(individual, "Fitness:", fitness(individual))

for generation in range(1, GENERATIONS + 1):
    new_population = []

    while len(new_population) < POPULATION_SIZE:
        parent1 = selection(population)
        parent2 = selection(population)

        child1, child2 = crossover(parent1, parent2)

        new_population.append(mutation(child1))
        if len(new_population) < POPULATION_SIZE:
            new_population.append(mutation(child2))

    population = new_population

    best = max(population, key=fitness)
    print(
        f"Generation {generation}: "
        f"Best x = {decode(best)}, "
        f"Fitness = {fitness(best)}"
    )

# Step 7: Display best solution
best = max(population, key=fitness)

print("\nFinal Result")
print("Best Chromosome:", best)
print("Best Solution (x):", decode(best))
print("Maximum Fitness:", fitness(best))
```
