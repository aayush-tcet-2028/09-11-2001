```python
print("McCulloch-Pitts (MCP) Neuron Model")
print("Implementation of AND, OR and XOR Logic Gates")
```

```python
def mcp_neuron(inputs, weights, threshold):
    """
    McCulloch-Pitts Neuron
    """

    weighted_sum = 0

    for i in range(len(inputs)):
        weighted_sum += inputs[i] * weights[i]

    if weighted_sum >= threshold:
        return 1
    else:
        return 0
```

```python
inputs = [
    [0,0],
    [0,1],
    [1,0],
    [1,1]
]

print("Input Combinations")
print(inputs)
```

```python
print("AND GATE")

print("Weights = [1,1]")
print("Threshold = 2\n")

print("X1\tX2\tOutput")

for x in inputs:
    y = mcp_neuron(x,[1,1],2)
    print(x[0],"\t",x[1],"\t",y)
```

```python
print("OR GATE")

print("Weights = [1,1]")
print("Threshold = 1\n")

print("X1\tX2\tOutput")

for x in inputs:
    y = mcp_neuron(x,[1,1],1)
    print(x[0],"\t",x[1],"\t",y)
```

```python
print("XOR GATE")

print("Single MCP neuron cannot solve XOR.")
print("Using a two-layer MCP Network.\n")

print("X1\tX2\tOutput")

for x in inputs:

    # Hidden Layer

    h1 = mcp_neuron(x,[1,-1],1)
    h2 = mcp_neuron(x,[-1,1],1)

    # Output Layer

    y = mcp_neuron([h1,h2],[1,1],1)

    print(x[0],"\t",x[1],"\t",y)
```

```python
print("Truth Table for AND")

print("--------------------")
print("X1 X2 Output")

for x in inputs:
    print(x[0],x[1],mcp_neuron(x,[1,1],2))


print("\nTruth Table for OR")

print("--------------------")
print("X1 X2 Output")

for x in inputs:
    print(x[0],x[1],mcp_neuron(x,[1,1],1))


print("\nTruth Table for XOR")

print("--------------------")
print("X1 X2 Output")

for x in inputs:

    h1 = mcp_neuron(x,[1,-1],1)
    h2 = mcp_neuron(x,[-1,1],1)

    y = mcp_neuron([h1,h2],[1,1],1)

    print(x[0],x[1],y)
```

```python
print("Test MCP Neuron")

x1 = int(input("Enter First Input (0/1): "))
x2 = int(input("Enter Second Input (0/1): "))

gate = input("Choose Gate (AND/OR): ").upper()

if gate == "AND":
    result = mcp_neuron([x1,x2],[1,1],2)

elif gate == "OR":
    result = mcp_neuron([x1,x2],[1,1],1)

else:
    result = "Invalid Gate"

print("\nResult =", result)
```
